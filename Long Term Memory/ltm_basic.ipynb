```python
from langgraph.store.memory import InMemoryStore
```

```python
# creating object of the in-memory store
store = InMemoryStore()
```

```python
# creating a namespace 
# Namespace is nothing but a folder inside a memory jaha humlog memory organise hoga
# Ek folder banaya User/u1 mtlb Parent folder users and uske andar u1
namespace = ("user","u1") # namespace tuple k form mai create hota hai
```

```python
# Creating memory
# for that we use Put method from the Basestore jo naya memory create karta hai namespcae mai
# Put(namespace, unique_key, value to store)

store.put(namespace,"1",{"data":"User like pizza"})
store.put(namespace,"2",{"data":"User prefers dark mode"})
```

```python
# another name space
namespace2 = ("user","u2")
# add memories
store.put(namespace2,"1",{"data":"User like Pasta"})
store.put(namespace2,"2",{"data":"User prefers grid style navigation"})
```

```python
# Reetrival of Memeory
# for that we use get method 
# get(namespace, unique_key)

store.get(namespace,"1")
```

```python
# Retriving all memories
# we user search(namespace)

items = store.search(namespace2)
for item in items:
    print(item.value)
```

```python
from langchain_ollama import OllamaEmbeddings
from dotenv import load_dotenv
load_dotenv()

model = OllamaEmbeddings(model="embeddinggemma:300m")
```

```python
store = InMemoryStore(index={"embed": model, "dims":1536})
```

```python

store.put(namespace, "1", {"data": "User prefers concise answers over long explanations"})
store.put(namespace, "2", {"data": "User likes examples in Python"})
store.put(namespace, "3", {"data": "User usually works late at night"})
store.put(namespace, "4", {"data": "User prefers dark mode in applications"})
store.put(namespace, "5", {"data": "User is learning machine learning"})
store.put(namespace, "6", {"data": "User dislikes overly theoretical explanations"})
store.put(namespace, "7", {"data": "User prefers step-by-step reasoning"})
store.put(namespace, "8", {"data": "User is based in India"})
store.put(namespace, "9", {"data": "User likes real-world analogies"})
store.put(namespace, "10", {"data": "User prefers bullet points over paragraphs"})
```

```python
items = store.search(namespace, query="What is the user currently learning", limit=1)

for item in items:
    print(item.value)
```

```python
items = store.search(namespace, query="what are user's preferences", limit=4)

for item in items:
    print(item.value)
```
